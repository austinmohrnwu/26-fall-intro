### Scenario 1

In the census table below, visualize the way the 2010 base estimate changes as age increases.

```python
census2010 = Table.read_table('nc-est2019-agesex-res.csv').where('SEX', 0).where('AGE', are.below(999)).select('AGE', 'ESTIMATESBASE2010')
census2010.show(3)
```

### Solution 1

A line plot is most appropriate since we have numerical input (age) and the output (2010 estimate) is a function of the input. (That is, each input specifies a single output.)

```python
census2010.plot('AGE', 'ESTIMATESBASE2010')
```

### Scenario 2

In the film table below, visualize the correlation (if any) between year of release and rating.

```python
movies = Table.read_table('imdb.csv')
movies.show(3)
```

### Solution 2

A scatter plot is most appropriate since we have numerical input (year) and numerical output (rating), but the output is not a function of the input. That is, each year will have many ratings associated with it (one for each movie released that year).

```python
movies.scatter('Year', 'Rating')
```

### Scenario 3

In the film table below, visualize the number of movies released by each studio.

```python
movies2 = Table.read_table('top_movies_2017.csv')
movies2.show(3)
```

### Solution 3

A bar chart is most appropriate since we have a categorical input (studio) and want to count the frequency with which each appears in the table.

```python
studio_counts = movies2.group('Studio') # must group first to get the frequencies
studio_counts.show(3)

studio_counts.barh('Studio') # plot is generated using the counts table
```

### Scenario 4

In the films table below, visualize the distribution of the votes cast.

```python
movies = Table.read_table('imdb.csv')
movies.show(3)
```

### Solution 4

A histogram is most appropriate since we have a numerical statistic for each movie and we want to visualize it's distribtion (that is, how the number of votes cast is spread out).

```python
movies.hist('Votes', bins = 10) # creates 10 equal-sized bins
```
